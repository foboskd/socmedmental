## 2. Анализ выбросов

Для обнаружения выбросов используем метод межквартильного размаха (IQR). Значения, выходящие за пределы [Q1 - 1.5*IQR, Q3 + 1.5*IQR], считаются выбросами.

```python
# Анализ выбросов методом IQR
numeric_vars = ['age', 'daily_screen_time_min', 'social_media_time_min',
                'negative_interactions_count', 'positive_interactions_count',
                'sleep_hours', 'physical_activity_min', 'anxiety_level',
                'stress_level', 'mood_level']

outlier_summary = []
for col in numeric_vars:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    n_outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    outlier_summary.append({
        'Признак': col,
        'Мин': df[col].min(),
        'Q1': Q1,
        'Медиана': df[col].median(),
        'Q3': Q3,
        'Макс': df[col].max(),
        'Нижняя граница': round(lower, 2),
        'Верхняя граница': round(upper, 2),
        'Выбросы': n_outliers,
        '% выбросов': round(n_outliers / len(df) * 100, 2)
    })

outlier_df = pd.DataFrame(outlier_summary)
outlier_df
```

**Выводы по выбросам для предобработки:**

- **Решение:** обработка выбросов не производится, так как данные синтетические и не содержат аномальных значений.

### Кодирование категориальных переменных

Для корректной работы с целевой переменной используем отдельный LabelEncoder, чтобы иметь возможность декодировать предсказания обратно в читаемые названия классов.

```python
# Кодирование признаков и целевой переменной.
# Категориальные признаки (gender, platform) — one-hot через get_dummies,
# чтобы для SVM/KNN/нейросети не возникал искусственный порядок категорий
# (LabelEncoder создаёт порядок 0<1<2, который для номинальных категорий
# искажает метрики расстояния и градиенты).
# Целевая mental_state остаётся через LabelEncoder (sklearn ожидает int-классы).
df = pd.get_dummies(df, columns=['gender', 'platform'], drop_first=False)

le_target = LabelEncoder()
df['mental_state'] = le_target.fit_transform(df['mental_state'])

print("Классы целевой переменной:",
      dict(zip(le_target.classes_, le_target.transform(le_target.classes_))))
print(f"\nКоличество признаков после get_dummies: {df.shape[1] - 1}")
print(f"gender_*  : {[c for c in df.columns if c.startswith('gender_')]}")
print(f"platform_*: {[c for c in df.columns if c.startswith('platform_')]}")
```

### Балансировка обучающей выборки методом SMOTETomek

Используем комбинацию SMOTE + Tomek Links:
- **SMOTE** создаёт синтетические примеры для миноритарных классов
- **Tomek Links** удаляет пограничные/шумные точки, которые могли возникнуть при оверсемплинге

**Ограничение степени оверсемплинга (×3):** в реальной практике увеличение класса более чем в несколько раз приводит к тому, что синтетические точки создаются интерполяцией между ближайшими соседями и модель выучивает искусственные паттерны, не обобщающиеся на реальные данные.

В нашем датасете большинство признаков имеют низкую кардинальность (3 значения у gender, 7 у platform, 4–5 у Likert-шкал anxiety/stress/mood, 6.4–8.3 с шагом 0.1 у sleep_hours). Из-за этого SMOTE между ближайшими соседями миноритарного класса часто порождает строки, идентичные уже существующим. Поэтому ограничиваем увеличение каждого класса **максимум в 3 раза** от исходного размера и опираемся на `class_weight='balanced'` в моделях, чтобы компенсировать остаточный дисбаланс без дальнейшей генерации синтетики.

```python
# ДИАГНОСТИЧЕСКАЯ балансировка — для отображения распределения после SMOTETomek.
# В обучении не участвует: модели обучаются через imblearn-Pipeline, где SMOTETomek
# вызывается внутри каждого CV-фолда (см. раздел «Подбор гиперпараметров»).
# Проверка качества синтетики (следующая ячейка) использует свои SMOTE+Tomek шаги
# отдельно — для позиционной идентификации синтетических строк.

majority_count = y_train.value_counts().max()
target_counts = {}
MAX_OVERSAMPLE_RATIO = 3   # уменьшено с 5 (см. markdown выше — низкая кардинальность фич)

for cls, cnt in y_train.value_counts().items():
    if cnt < majority_count:
        target_counts[cls] = min(int(cnt * MAX_OVERSAMPLE_RATIO), majority_count)
    else:
        target_counts[cls] = cnt

print(f"Исходное распределение: {dict(y_train.value_counts())}")
print(f"Целевое распределение для SMOTE (макс. x{MAX_OVERSAMPLE_RATIO}): {target_counts}")

smote_tomek_diag = SMOTETomek(
    smote=SMOTE(sampling_strategy=target_counts, random_state=42, k_neighbors=5),
    random_state=42
)
X_train_bal, y_train_bal = smote_tomek_diag.fit_resample(X_train, y_train)

print(f"\nРазмер до балансировки: {X_train.shape[0]}")
print(f"Размер после балансировки: {X_train_bal.shape[0]}")
print(f"\nРаспределение классов после балансировки:")
print(pd.Series(y_train_bal).value_counts())
```

**Примечание:** этот блок балансировки — **исключительно диагностический**, он нужен только для следующей ячейки (проверка свойств синтетических точек). В обучении моделей используется `imblearn.pipeline.Pipeline`, внутри которого `SMOTETomek` применяется **отдельно к каждому train-фолду кросс-валидации** — это исключает утечку, описанную в замечании рецензента №2.

Степень оверсемплинга ограничена множителем x5 от исходного размера каждого класса.


### Проверка качества синтетических данных (SMOTE)

После применения SMOTE необходимо убедиться, что сгенерированные точки не являются «копиями» друг друга и образуют разнообразное распределение. Для этого:
1. Сравниваем статистики (среднее, стд) оригинальных и синтетических данных
2. Визуализируем распределение через PCA

```python
# Проверка качества сгенерированных SMOTE-точек.
#
# Старая логика: hash-based детекция (строка-синтетика, если её хеш не совпал ни с одним
# хешом в исходном X_train). На наших данных это давало ОШИБКУ: feature space очень
# маленький (gender 0/1/2, platform 0..6, integer Likert 4-5 уровней), и SMOTE между
# близкими/одинаковыми соседями миноритарного класса генерирует строки, чья комбинация
# фич случайно совпадает с уже существующей мажоритарной строкой. Hash-сравнение
# проверяет только фичи (без класса), поэтому такие синтетические строки помечались
# "оригиналом" и счётчик "originals after Tomek" получался > исходного размера
# (вплоть до отрицательного "удалено Tomek Links").
#
# Новая логика: позиционная идентификация через документированную imblearn-конвенцию.
# SMOTE.fit_resample возвращает оригиналы первыми (первые len(X_train) строк),
# синтетика идёт в хвосте. Затем TomekLinks.sample_indices_ даёт индексы строк,
# выживших после Tomek, — этого достаточно для точного маппинга.

from sklearn.decomposition import PCA
from imblearn.over_sampling import SMOTE
from imblearn.under_sampling import TomekLinks

smote_diag = SMOTE(sampling_strategy=target_counts, random_state=42, k_neighbors=5)
X_smote, y_smote = smote_diag.fit_resample(X_train, y_train)

n_orig = len(X_train)
is_synth_after_smote = np.zeros(len(X_smote), dtype=bool)
is_synth_after_smote[n_orig:] = True   # imblearn: оригиналы первыми, синтетика — в хвосте

tomek_diag = TomekLinks()
X_bal_arr, y_bal_arr_raw = tomek_diag.fit_resample(X_smote, y_smote)
kept_idx = tomek_diag.sample_indices_   # индексы в X_smote, выжившие после Tomek
is_synthetic = is_synth_after_smote[kept_idx]

bal_df = pd.DataFrame(X_bal_arr, columns=X_train.columns).reset_index(drop=True)
X_original_kept = bal_df[~is_synthetic]
X_synthetic    = bal_df[ is_synthetic]
y_bal_arr = np.asarray(y_bal_arr_raw)
y_original_kept = y_bal_arr[~is_synthetic]
y_synthetic    = y_bal_arr[ is_synthetic]

# Сanity-check: оригиналов после Tomek физически не может стать больше, чем было.
assert len(X_original_kept) <= len(X_train), \
    f"Bug: original kept ({len(X_original_kept)}) > original ({len(X_train)})"

print("=== Сравнение оригинальных и синтетических данных (позиционная идентификация) ===\n")
print(f"Оригинальных точек до балансировки: {len(X_train)}")
print(f"Оригинальных точек, переживших Tomek: {len(X_original_kept)}  "
      f"(удалено Tomek Links: {len(X_train) - len(X_original_kept)})")
print(f"Синтетических точек, созданных SMOTE и переживших Tomek: {len(X_synthetic)}")
print(f"Всего после балансировки: {len(bal_df)}")

if len(X_synthetic) > 0:
    print("\nСреднее значение признаков (первые 5 числовых колонок):")
    print(f"  Оригинальные:   {X_original_kept.mean().values[:5].round(2)}")
    print(f"  Синтетические:  {X_synthetic.mean().values[:5].round(2)}")

    print("\nСтандартное отклонение признаков (первые 5):")
    print(f"  Оригинальные:   {X_original_kept.std().values[:5].round(2)}")
    print(f"  Синтетические:  {X_synthetic.std().values[:5].round(2)}")

    n_dup_synth = X_synthetic.duplicated().sum()
    print(f"\nДубликаты среди синтетических точек: {n_dup_synth} из {len(X_synthetic)} "
          f"({n_dup_synth/len(X_synthetic)*100:.1f}%)")

    print("\nРаспределение классов среди синтетических точек:")
    print(pd.Series(y_synthetic).value_counts().to_dict())

# PCA-визуализация
pca = PCA(n_components=2)
X_pca = pca.fit_transform(bal_df.values.astype(float))

fig, axes = plt.subplots(1, 2, figsize=(16, 6))
for cls in np.unique(y_bal_arr):
    mask = y_bal_arr == cls
    axes[0].scatter(X_pca[mask, 0], X_pca[mask, 1], alpha=0.3, s=10, label=f'Класс {cls}')
axes[0].set_title('PCA: распределение по классам после балансировки')
axes[0].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
axes[0].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
axes[0].legend()

axes[1].scatter(X_pca[~is_synthetic, 0], X_pca[~is_synthetic, 1],
                alpha=0.3, s=10, label='Оригинальные (пережили Tomek)', color='blue')
axes[1].scatter(X_pca[is_synthetic, 0], X_pca[is_synthetic, 1],
                alpha=0.3, s=10, label='Синтетические (SMOTE)', color='red')
axes[1].set_title('PCA: оригинальные vs синтетические (позиционная идентификация)')
axes[1].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
axes[1].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
axes[1].legend()

plt.tight_layout()
plt.show()

print("\nВывод: позиционная идентификация даёт корректный счётчик "
      "(оригиналов после Tomek <= исходного размера). Доля дубликатов среди синтетики "
      "отражает свойство данных (низкая кардинальность дискретных признаков → SMOTE между "
      "соседями миноритарного класса часто порождает повторяющиеся комбинации). "
      "Этот эффект ограничивается уменьшением oversampling-ratio (см. ячейку с MAX_OVERSAMPLE_RATIO) "
      "и компенсируется class_weight='balanced' в моделях.")
```

# Построение и сравнение моделей

### Набор метрик

При дисбалансе классов оцениваем модели по расширенному набору метрик:
- **Precision** — доля истинно положительных среди всех предсказанных положительных
- **Recall** — доля найденных положительных из всех реально положительных
- **F1-score** — гармоническое среднее precision и recall
- **ROC-AUC (macro)** — площадь под ROC-кривой, усреднённая по классам
- **Confusion Matrix** — матрица ошибок для визуального анализа
- **Accuracy** — общая точность (для справки, но не как основная метрика)

### Подбор гиперпараметров (GridSearchCV)

Для корректного сравнения моделей необходимо подобрать оптимальные гиперпараметры для каждой модели с помощью кросс-валидации. Без GridSearch нельзя утверждать, что модели работают оптимально — ручной выбор параметров может существенно занижать или завышать результаты.

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

# imblearn-Pipeline: SMOTETomek + StandardScaler + classifier.
# При вызове grid.fit(X_train, y_train) КАЖДЫЙ CV-фолд будет независимо
# масштабирован и сбалансирован — утечка через SMOTE исключена.

def make_pipe(estimator):
    return ImbPipeline([
        ("scaler", StandardScaler()),
        ("smote", SMOTETomek(
            smote=SMOTE(sampling_strategy=target_counts, random_state=42, k_neighbors=5),
            random_state=42
        )),
        ("clf", estimator),
    ])

param_grids = {
    'Logistic Regression': {
        'estimator': LogisticRegression(class_weight='balanced', max_iter=1000),
        'params': {
            'clf__C': [0.01, 0.1, 1.0, 10.0],
            'clf__solver': ['lbfgs', 'liblinear'],
        },
    },
    'Decision Tree': {
        'estimator': DecisionTreeClassifier(class_weight='balanced', random_state=42),
        'params': {
            'clf__max_depth': [5, 10, 15, 20, None],
            'clf__min_samples_leaf': [1, 3, 5, 10],
            'clf__min_samples_split': [2, 5, 10],
        },
    },
    'Random Forest': {
        'estimator': RandomForestClassifier(class_weight='balanced', random_state=42),
        'params': {
            'clf__n_estimators': [50, 100, 200],
            'clf__max_depth': [10, 15, 20, None],
            'clf__min_samples_leaf': [1, 3, 5],
        },
    },
    'SVM': {
        'estimator': SVC(class_weight='balanced', probability=True, random_state=42),
        'params': {
            'clf__C': [0.1, 1.0, 10.0],
            'clf__kernel': ['rbf', 'linear'],
            'clf__gamma': ['scale', 'auto'],
        },
    },
    'KNN': {
        'estimator': KNeighborsClassifier(),
        'params': {
            'clf__n_neighbors': [3, 5, 7, 11, 15],
            'clf__weights': ['uniform', 'distance'],
            'clf__metric': ['euclidean', 'manhattan'],
        },
    },
}

best_models = {}
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

for name, cfg in param_grids.items():
    print(f"\n{'='*60}\n  GridSearchCV: {name}\n{'='*60}")
    grid = GridSearchCV(
        estimator=make_pipe(cfg['estimator']),
        param_grid=cfg['params'],
        cv=cv,
        scoring='f1_macro',
        n_jobs=-1,
        verbose=0,
    )
    # Обучаем на НЕсбалансированном и НЕотскейленном train — всё делает pipeline per-fold
    grid.fit(X_train, y_train)
    best_models[name] = {
        'pipeline': grid.best_estimator_,
        'best_params': grid.best_params_,
        'best_score': grid.best_score_,
    }
    print(f"  Лучшие параметры: {grid.best_params_}")
    print(f"  Лучший F1 (macro) на CV: {grid.best_score_:.4f}")
```

### Итоговые модели с оптимальными гиперпараметрами

```python
# Итоговые пайплайны с оптимальными гиперпараметрами из GridSearchCV
models = {name: cfg['pipeline'] for name, cfg in best_models.items()}

print("Итоговые пайплайны (scaler -> SMOTETomek -> clf) с оптимальными параметрами:")
for name, cfg in best_models.items():
    print(f"\n  {name}:")
    print(f"    Параметры: {cfg['best_params']}")
    print(f"    CV F1 (macro): {cfg['best_score']:.4f}")
```

### Нейронная сеть (Keras) — подбор гиперпараметров по сетке

Для нейронной сети также выполняем GridSearchCV через обёртку KerasClassifier (scikeras).
Подбираем:
- Количество скрытых слоёв и нейронов
- Функцию активации
- Оптимизатор
- Размер батча и количество эпох

```python
import os
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.regularizers import l2
from scikeras.wrappers import KerasClassifier

# Ограничим число потоков внутри TF, чтобы параллельные воркеры joblib не
# перетягивали друг у друга CPU. Итого: joblib запускает N процессов,
# каждый TF-процесс использует 1–2 потока. Итог ~ N_CORES суммарно.
os.environ.setdefault("TF_CPP_MIN_LOG_LEVEL", "2")
tf.config.threading.set_intra_op_parallelism_threads(1)
tf.config.threading.set_inter_op_parallelism_threads(1)

n_features = X_train.shape[1]
n_classes = len(np.unique(y_train))

def create_model(n_hidden_layers=1, neurons=16, activation='relu', optimizer='adam'):
    model = Sequential()
    model.add(Input(shape=(n_features,)))
    model.add(Dense(neurons, activation=activation, kernel_regularizer=l2(0.01)))
    model.add(Dropout(0.3))
    for _ in range(n_hidden_layers - 1):
        model.add(Dense(neurons, activation=activation, kernel_regularizer=l2(0.01)))
        model.add(Dropout(0.3))
    model.add(Dense(n_classes, activation='softmax'))
    model.compile(
        loss='sparse_categorical_crossentropy',
        optimizer=optimizer,
        metrics=['accuracy']
    )
    return model

# При n_jobs != 1 логи Keras из воркеров перемешиваются, поэтому verbose=0.
# Прогресс будем видеть из joblib: "Done N out of M tasks".
keras_clf = KerasClassifier(model=create_model, verbose=0)
nn_pipe = ImbPipeline([
    ("scaler", StandardScaler()),
    ("smote", SMOTETomek(
        smote=SMOTE(sampling_strategy=target_counts, random_state=42, k_neighbors=5),
        random_state=42
    )),
    ("clf", keras_clf),
])

param_grid_nn = {
    'clf__model__n_hidden_layers': [2, 4],
    'clf__model__neurons': [32, 64],
    'clf__model__activation': ['relu', 'tanh'],
    'clf__model__optimizer': ['adam', 'rmsprop'],
    'clf__batch_size': [16, 32],
    'clf__epochs': [50],
}

grid_nn = GridSearchCV(estimator=nn_pipe, param_grid=param_grid_nn,
                       cv=2, scoring='f1_macro',
                       n_jobs=10, verbose=2)
grid_nn_result = grid_nn.fit(X_train, y_train)

print(f"\nЛучший F1 (macro) на CV: {grid_nn_result.best_score_:.4f}")
print(f"Лучшие параметры: {grid_nn_result.best_params_}")
best_nn_pipeline = grid_nn_result.best_estimator_
```

```python
# Визуализация: лучшие параметры нейронной сети
print("Лучшие параметры нейронной сети (GridSearchCV):")
for param, value in grid_nn_result.best_params_.items():
    print(f"  {param}: {value}")
print(f"Лучший F1 (macro) на кросс-валидации: {grid_nn_result.best_score_:.4f}")
```

# Оценка моделей

```python
def evaluate_pipeline(pipeline, X_tr, y_tr, X_te, y_te, model_name, already_fit=True):
    """Оценивает imblearn-Pipeline (SMOTE + scaler + clf).
    Если already_fit=True (по умолчанию), использует grid.best_estimator_,
    который уже обучен на полном train через GridSearchCV.
    """
    if not already_fit:
        pipeline.fit(X_tr, y_tr)
    y_pred = pipeline.predict(X_te)
    try:
        y_proba = pipeline.predict_proba(X_te)
    except Exception:
        y_proba = None

    acc = float(np.mean(y_pred == y_te))
    f1_macro = f1_score(y_te, y_pred, average='macro')
    f1_weighted = f1_score(y_te, y_pred, average='weighted')
    prec_macro = precision_score(y_te, y_pred, average='macro', zero_division=0)
    rec_macro = recall_score(y_te, y_pred, average='macro', zero_division=0)
    roc_auc = None
    if y_proba is not None:
        try:
            roc_auc = roc_auc_score(y_te, y_proba, multi_class='ovr', average='macro')
        except Exception:
            roc_auc = None

    return {
        'Model': model_name,
        'Accuracy': round(acc, 4),
        'Precision (macro)': round(prec_macro, 4),
        'Recall (macro)': round(rec_macro, 4),
        'F1 (macro)': round(f1_macro, 4),
        'F1 (weighted)': round(f1_weighted, 4),
        'ROC-AUC (macro)': round(roc_auc, 4) if roc_auc else 'N/A',
        'y_pred': y_pred,
        'y_proba': y_proba,
    }
```

```python
results = []
predictions = {}

for name, pipeline in models.items():
    # pipeline уже обучен на train через GridSearchCV — не переобучаем
    res = evaluate_pipeline(pipeline, X_train, y_train, X_test, y_test, name, already_fit=True)
    results.append(res)
    predictions[name] = res
    print(f"{'='*60}\n  {name}\n{'='*60}")
    print(classification_report(y_test, res['y_pred'],
          target_names=le_target.classes_, zero_division=0))

# Neural network pipeline
nn_res = evaluate_pipeline(best_nn_pipeline, X_train, y_train, X_test, y_test,
                            'Neural Network', already_fit=True)
results.append(nn_res)
predictions['Neural Network'] = nn_res
print(f"{'='*60}\n  Neural Network\n{'='*60}")
print(classification_report(y_test, nn_res['y_pred'],
      target_names=le_target.classes_, zero_division=0))
```

# Learning curve и ablation-анализ

Два простых теста, которые подтверждают «формульную» природу таргета:
- **Learning curve** — если при 10% данных метрика уже около 1.0, значит модель не нуждается в большом объёме обучения, достаточно увидеть формулу.
- **Ablation** — уберём топ-признаки и посмотрим, как упадёт F1. Если падение резкое — таргет действительно детерминирован этими фичами.


```python
# B.1 Learning curve для Random Forest
from sklearn.ensemble import RandomForestClassifier

rf_lc_estimator = ImbPipeline([
    ("scaler", StandardScaler()),
    ("smote", SMOTETomek(
        smote=SMOTE(sampling_strategy=target_counts, random_state=42, k_neighbors=5),
        random_state=42)),
    ("clf", RandomForestClassifier(
        **{k.replace("clf__", ""): v for k, v in best_models['Random Forest']['best_params'].items()},
        class_weight='balanced', random_state=42)),
])

train_sizes, train_scores, val_scores = learning_curve(
    rf_lc_estimator, X_train, y_train,
    train_sizes=np.linspace(0.1, 1.0, 6),
    cv=3, scoring='f1_macro', n_jobs=-1, random_state=42
)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(train_sizes, train_scores.mean(axis=1), 'o-', label='train F1 (macro)')
ax.plot(train_sizes, val_scores.mean(axis=1), 's-', label='val F1 (macro)')
ax.set_xlabel('Размер тренировочной выборки')
ax.set_ylabel('F1 (macro)')
ax.set_title('Learning curve: Random Forest')
ax.set_ylim(0, 1.05)
ax.legend()
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nval F1 (macro) по размерам train:")
for s, m in zip(train_sizes, val_scores.mean(axis=1)):
    print(f"  train_size={s:5.0f}  val_F1={m:.4f}")

```

```python
# B.2 Ablation: удалить топ-1 и топ-3 признака по важности, переобучить, сравнить F1
top_features = feat_importance.sort_values(ascending=False).index.tolist()
ablation_results = []

for k_drop in [0, 1, 3]:
    drop_cols = top_features[:k_drop]
    X_tr_ab = X_train.drop(columns=drop_cols)
    X_te_ab = X_test.drop(columns=drop_cols)

    rf_params = {kk.replace("clf__", ""): vv for kk, vv in best_models['Random Forest']['best_params'].items()}
    ab_pipe = ImbPipeline([
        ("scaler", StandardScaler()),
        ("smote", SMOTETomek(random_state=42)),
        ("clf", RandomForestClassifier(**rf_params, class_weight='balanced', random_state=42)),
    ])
    ab_pipe.fit(X_tr_ab, y_train)
    y_pred_ab = ab_pipe.predict(X_te_ab)
    f1_ab = f1_score(y_test, y_pred_ab, average='macro')
    ablation_results.append({
        'dropped_features': drop_cols if drop_cols else '(none)',
        'k_dropped': k_drop,
        'F1_macro_test': round(f1_ab, 4),
    })

print(pd.DataFrame(ablation_results).to_string(index=False))
print("\nРезкое падение F1 при удалении топ-признаков => таргет действительно")
print("детерминирован небольшим подмножеством фич (свойство синтетики).")

```

# Перенос модели на реальный датасет `smmh.csv`

До сих пор все метрики считались на test-подвыборке из **того же синтетического датасета**. Чтобы честно оценить прикладную ценность модели, применим её к независимому реальному опроснику: https://www.kaggle.com/datasets/souvikahmed071/social-media-and-mental-health (`smmh.csv`, ~481 реальный респондент, апрель 2022).

Прямого соответствия схем нет — сделаем маппинг и построим эвристический псевдо-таргет по индикаторам дистресса из анкеты.


```python
# D.1 Загрузка и первичная очистка smmh.csv
import os

SMMH_PATH = 'smmh.csv'
if not os.path.exists(SMMH_PATH):
    raise FileNotFoundError(
        f"Файл {SMMH_PATH} не найден. Скачайте его с "
        "https://www.kaggle.com/datasets/souvikahmed071/social-media-and-mental-health "
        "и положите рядом с notebook."
    )

smmh = pd.read_csv(SMMH_PATH)
print(f"smmh.csv: {smmh.shape}")

# Отбросить ненужные колонки, оставить только связанные с нашими фичами
smmh = smmh.rename(columns={
    '1. What is your age?': 'q_age',
    '2. Gender': 'q_gender',
    '6. Do you use social media?': 'q_use_sm',
    '7. What social media platforms do you commonly use?': 'q_platforms',
    '8. What is the average time you spend on social media every day?': 'q_sm_time',
    '11. Do you feel restless if you haven\'t used Social media in a while?': 'q_restless',
    '12. On a scale of 1 to 5, how easily distracted are you?': 'q_distracted',
    '13. On a scale of 1 to 5, how much are you bothered by worries?': 'q_worries',
    '14. Do you find it difficult to concentrate on things?': 'q_concentrate',
    '16. Following the previous question, how do you feel about these comparisons, generally speaking?': 'q_compare_feel',
    '17. How often do you look to seek validation from features of social media?': 'q_seek_validation',
    '18. How often do you feel depressed or down?': 'q_depressed',
    '20. On a scale of 1 to 5, how often do you face issues regarding sleep?': 'q_sleep_issues',
})

smmh = smmh[smmh['q_use_sm'].astype(str).str.strip() == 'Yes'].copy()
print(f"После фильтра 'use social media = Yes': {smmh.shape}")
smmh = smmh.dropna(subset=['q_age', 'q_sm_time']).copy()
print(f"После удаления NA по возрасту и времени соцсетей: {smmh.shape}")

```

```python
# D.2 Маппинг 21-колоночного опросника в фичи нашей модели (с one-hot для gender/platform)
def normalize_gender(g):
    g = str(g).strip().lower()
    if g.startswith('male'): return 'Male'
    if g.startswith('female'): return 'Female'
    return 'Other'

# Список тренировочных платформ восстанавливаем из имён dummy-колонок X_train
TRAIN_PLATFORMS = {c.replace('platform_', '') for c in X_train.columns
                   if c.startswith('platform_')}

def pick_platform(s):
    if pd.isna(s):
        return 'Instagram'
    for p in [x.strip() for x in str(s).split(',')]:
        if p in TRAIN_PLATFORMS:
            return p
    return 'Instagram'

time_midpoint = {
    'Less than an Hour':     30,
    'Between 1 and 2 hours': 90,
    'Between 2 and 3 hours': 150,
    'Between 3 and 4 hours': 210,
    'Between 4 and 5 hours': 270,
    'More than 5 hours':     360,
}
sleep_map = {1: 8.3, 2: 7.8, 3: 7.3, 4: 6.9, 5: 6.4}
mood_map  = {1: 7,   2: 6,   3: 6,   4: 5,   5: 4}

def neg_map(v):
    if v <= 2: return 0
    if v == 3: return 1
    return 2

train_phys_median = float(X_train['physical_activity_min'].median())

def adapt_smmh(df_raw):
    df = df_raw.copy()
    df['age'] = df['q_age'].astype(float).clip(13, 69).astype(int)
    df['gender'] = df['q_gender'].apply(normalize_gender)
    df['platform'] = df['q_platforms'].apply(pick_platform)
    df['social_media_time_min'] = df['q_sm_time'].map(time_midpoint).fillna(150).astype(int)
    df['daily_screen_time_min'] = (df['social_media_time_min'] * 2).clip(140, 520).astype(int)
    df['anxiety_level'] = df['q_worries'].clip(1, 4).astype(int)
    df['stress_level'] = (4 + df[['q_restless', 'q_distracted', 'q_worries', 'q_concentrate']].mean(axis=1).round()).clip(5, 9).astype(int)
    df['mood_level'] = df['q_depressed'].map(mood_map).clip(4, 7).astype(int)
    df['sleep_hours'] = df['q_sleep_issues'].map(sleep_map).astype(float)
    df['negative_interactions_count'] = df['q_compare_feel'].apply(neg_map).astype(int)
    df['positive_interactions_count'] = (6 - df['q_seek_validation']).clip(0, 4).astype(int)
    df['physical_activity_min'] = train_phys_median

    # Псевдо-таргет (тот же эвристический способ)
    score = (df['q_restless'] + df['q_worries'] + df['q_concentrate']
             + df['q_depressed'] + df['q_sleep_issues'])
    pseudo = np.where(score <= 10, 'Healthy', np.where(score <= 17, 'At_Risk', 'Stressed'))

    # Базовый набор фич, ещё со строковыми gender/platform
    raw_features = pd.DataFrame({
        'age': df['age'],
        'gender': df['gender'],
        'platform': df['platform'],
        'daily_screen_time_min': df['daily_screen_time_min'],
        'social_media_time_min': df['social_media_time_min'],
        'negative_interactions_count': df['negative_interactions_count'],
        'positive_interactions_count': df['positive_interactions_count'],
        'sleep_hours': df['sleep_hours'],
        'physical_activity_min': df['physical_activity_min'],
        'anxiety_level': df['anxiety_level'],
        'stress_level': df['stress_level'],
        'mood_level': df['mood_level'],
    })

    # one-hot и выравнивание колонок под тренировочный feature space
    out = pd.get_dummies(raw_features, columns=['gender', 'platform'])
    out = out.reindex(columns=X_train.columns, fill_value=0)

    y_pseudo = le_target.transform(pseudo)
    return out, y_pseudo, pseudo

X_smmh, y_smmh_pseudo, y_smmh_labels = adapt_smmh(smmh)
print(f"Адаптированный smmh: {X_smmh.shape}")
print(f"Колонки совпадают с X_train: {list(X_smmh.columns) == list(X_train.columns)}")
print("\nРаспределение псевдо-таргета:")
print(pd.Series(y_smmh_labels).value_counts().to_dict())
print("\nПервые строки:")
print(X_smmh.head())
```

```python
# D.4 Альтернативная "совместимая" модель — без прокси для positive_interactions_count
# и physical_activity_min (они в smmh.csv не наблюдаются напрямую).
# После get_dummies в COMPATIBLE_FEATURES попадают все колонки, кроме двух прокси-фич.
EXCLUDE = {'positive_interactions_count', 'physical_activity_min'}
COMPATIBLE_FEATURES = [c for c in X_train.columns if c not in EXCLUDE]

X_train_compat = X_train[COMPATIBLE_FEATURES]
X_test_compat = X_test[COMPATIBLE_FEATURES]
X_smmh_compat = X_smmh[COMPATIBLE_FEATURES]

# Обучим RF (как лучшая модель) на подмножестве фич через тот же Pipeline
rf_params_compat = {k.replace("clf__", ""): v for k, v in best_models['Random Forest']['best_params'].items()}
pipe_compat = ImbPipeline([
    ("scaler", StandardScaler()),
    ("smote", SMOTETomek(
        smote=SMOTE(sampling_strategy=target_counts, random_state=42, k_neighbors=5),
        random_state=42)),
    ("clf", RandomForestClassifier(**rf_params_compat, class_weight='balanced', random_state=42)),
])
pipe_compat.fit(X_train_compat, y_train)

# Тест-метрики на нашей выборке
y_pred_test_compat = pipe_compat.predict(X_test_compat)
f1_test_compat = f1_score(y_test, y_pred_test_compat, average='macro')
acc_test_compat = float(np.mean(y_pred_test_compat == y_test))
print(f"Compat-модель на нашем test ({len(COMPATIBLE_FEATURES)} фич): "
      f"acc={acc_test_compat:.4f}, F1-macro={f1_test_compat:.4f}")

# Перенос на smmh
y_pred_smmh_compat = pipe_compat.predict(X_smmh_compat)
print("\nCompat-модель на smmh — classification_report vs псевдо-таргет:")
print(classification_report(y_smmh_pseudo, y_pred_smmh_compat,
      target_names=le_target.classes_, zero_division=0))

# Сравнительная таблица
compare = pd.DataFrame([
    {'Модель': f'{best_name} (полная, {X_train.shape[1]} фич)',
     'test F1-macro (синтетика)': round(best_models[best_name]['best_score'], 4),
     'smmh accuracy vs псевдо': round(float(np.mean(y_pred_smmh_full == y_smmh_pseudo)), 4),
     'smmh F1-macro vs псевдо': round(f1_score(y_smmh_pseudo, y_pred_smmh_full, average='macro'), 4)},
    {'Модель': f'RF compat ({len(COMPATIBLE_FEATURES)} фич, без прокси)',
     'test F1-macro (синтетика)': round(f1_test_compat, 4),
     'smmh accuracy vs псевдо': round(float(np.mean(y_pred_smmh_compat == y_smmh_pseudo)), 4),
     'smmh F1-macro vs псевдо': round(f1_score(y_smmh_pseudo, y_pred_smmh_compat, average='macro'), 4)},
])
print("\n=== Сравнение моделей на реальном датасете ===")
print(compare.to_string(index=False))
```

# Альтернативный прогон: обучение на `smmh.csv`

До этого места модель обучалась на синтетическом датасете и проверялась на реальных данных в режиме переноса — качество оказалось низким (F1-macro ≈ 0.32).

Попробуем **обратный сценарий**: тот же пайплайн (`StandardScaler → SMOTETomek → clf` с GridSearchCV), но обученный **на реальных данных smmh.csv**. Смотрим, что получится, когда модель обучается на тех же опросных признаках, на которых потом работает.

**Важное ограничение:** в `smmh.csv` нет готовой метки `mental_state`. Используем тот же эвристический псевдо-таргет по сумме индикаторов дистресса (Q11 restless + Q13 worries + Q14 concentrate + Q18 depressed + Q20 sleep → Healthy/At_Risk/Stressed). Это не клиническая метка, но хотя бы согласованная по датасету.

**Важное замечание про утечку:** признаки, из которых строится псевдо-таргет (Q11, Q13, Q14, Q18, Q20), **исключены из X** — иначе модель тривиально выучит обратную формулу. Оставляем только косвенные признаки: демография, время в соцсетях, платформа, отвлечения, сравнение с другими, seek_validation, interest_fluctuation.


> **Важно — характер этой проверки.**
>
> Целевая переменная `sm_target_str` ниже строится **эвристически** — это сумма пяти Likert-ответов (`q_restless + q_worries + q_concentrate + q_depressed + q_sleep_issues`) с порогами 10 и 17 для границ классов `Healthy / At_Risk / Stressed`. Сами признаки — реальные ответы респондентов, но **готовой клинической разметки `mental_state` в датасете smmh.csv нет**.
>
> Это означает: данный блок — **НЕ валидация модели на реальной размеченной задаче**, а проверка методологической согласованности пайплайна на реальных по природе данных. Метрики ниже следует читать как «модель учится восстанавливать собственный эвристический таргет», а не как доказательство клинической применимости.
>
> Для полноценной валидации потребовался бы датасет с разметкой `mental_state`, выставленной специалистом по принятой клинической шкале (PHQ-9, GAD-7 и т. п.).


```python
# SM.1 Загрузка smmh.csv и построение признаков/таргета (в переменных с префиксом sm_)
import os
sm_raw = pd.read_csv('smmh.csv')

sm_raw = sm_raw.rename(columns={
    '1. What is your age?': 'age',
    '2. Gender': 'gender',
    '3. Relationship Status': 'relationship',
    '4. Occupation Status': 'occupation',
    '6. Do you use social media?': 'use_sm',
    '7. What social media platforms do you commonly use?': 'platforms',
    '8. What is the average time you spend on social media every day?': 'sm_time',
    '9. How often do you find yourself using Social media without a specific purpose?': 'q_purpose',
    '10. How often do you get distracted by Social media when you are busy doing something?': 'q_distract_sm',
    "11. Do you feel restless if you haven't used Social media in a while?": 'q_restless',
    '12. On a scale of 1 to 5, how easily distracted are you?': 'q_distract',
    '13. On a scale of 1 to 5, how much are you bothered by worries?': 'q_worries',
    '14. Do you find it difficult to concentrate on things?': 'q_concentrate',
    '15. On a scale of 1-5, how often do you compare yourself to other successful people through the use of social media?': 'q_compare',
    '16. Following the previous question, how do you feel about these comparisons, generally speaking?': 'q_compare_feel',
    '17. How often do you look to seek validation from features of social media?': 'q_seek_validation',
    '18. How often do you feel depressed or down?': 'q_depressed',
    '19. On a scale of 1 to 5, how frequently does your interest in daily activities fluctuate?': 'q_interest_fluct',
    '20. On a scale of 1 to 5, how often do you face issues regarding sleep?': 'q_sleep_issues',
})

# Фильтры и чистка
sm_raw = sm_raw[sm_raw['use_sm'].astype(str).str.strip() == 'Yes'].copy()
sm_raw = sm_raw.dropna(subset=['age', 'sm_time']).copy()

def sm_norm_gender(g):
    g = str(g).strip().lower()
    if g.startswith('male'): return 'Male'
    if g.startswith('female'): return 'Female'
    return 'Other'

# Платформа — first match из whitelist (как в transfer-блоке)
sm_platforms_all = ['Instagram', 'Snapchat', 'Facebook', 'WhatsApp', 'TikTok', 'YouTube', 'Twitter']
def sm_pick_platform(s):
    if pd.isna(s): return 'Other'
    for p in [x.strip() for x in str(s).split(',')]:
        if p in sm_platforms_all: return p
    return 'Other'

sm_time_midpoint = {
    'Less than an Hour': 30, 'Between 1 and 2 hours': 90,
    'Between 2 and 3 hours': 150, 'Between 3 and 4 hours': 210,
    'Between 4 and 5 hours': 270, 'More than 5 hours': 360,
}

# Псевдо-таргет (см. markdown-disclaimer выше: построен эвристически из 5 Likert-индикаторов;
# это НЕ клиническая разметка, а проверка согласованности пайплайна)
sm_raw = sm_raw.dropna(subset=['q_restless', 'q_worries', 'q_concentrate', 'q_depressed', 'q_sleep_issues']).copy()
sm_score = (sm_raw['q_restless'] + sm_raw['q_worries'] + sm_raw['q_concentrate']
            + sm_raw['q_depressed'] + sm_raw['q_sleep_issues'])
sm_target_str = np.where(sm_score <= 10, 'Healthy',
                 np.where(sm_score <= 17, 'At_Risk', 'Stressed'))

print(f"smmh после очистки: {sm_raw.shape[0]} строк")
print(f"Распределение таргета: {pd.Series(sm_target_str).value_counts().to_dict()}")

# Признаки — ТОЛЬКО те, что НЕ участвуют в формуле таргета
# (исключаем q_restless, q_worries, q_concentrate, q_depressed, q_sleep_issues)
sm_features = pd.DataFrame({
    'age': sm_raw['age'].astype(int),
    'gender': sm_raw['gender'].apply(sm_norm_gender),
    'relationship': sm_raw['relationship'].fillna('Unknown'),
    'occupation': sm_raw['occupation'].fillna('Unknown'),
    'platform': sm_raw['platforms'].apply(sm_pick_platform),
    'sm_time_min': sm_raw['sm_time'].map(sm_time_midpoint).fillna(150).astype(int),
    'q_purpose': sm_raw['q_purpose'].fillna(sm_raw['q_purpose'].median()),
    'q_distract_sm': sm_raw['q_distract_sm'].fillna(sm_raw['q_distract_sm'].median()),
    'q_distract': sm_raw['q_distract'].fillna(sm_raw['q_distract'].median()),
    'q_compare': sm_raw['q_compare'].fillna(sm_raw['q_compare'].median()),
    'q_compare_feel': sm_raw['q_compare_feel'].fillna(sm_raw['q_compare_feel'].median()),
    'q_seek_validation': sm_raw['q_seek_validation'].fillna(sm_raw['q_seek_validation'].median()),
    'q_interest_fluct': sm_raw['q_interest_fluct'].fillna(sm_raw['q_interest_fluct'].median()),
}).reset_index(drop=True)

# One-hot для категорий вместо LabelEncoder — чтобы для SVM/KNN/нейросети не было
# искусственного порядка, который ухудшает метрики расстояния и градиенты.
sm_features = pd.get_dummies(sm_features,
    columns=['gender', 'relationship', 'occupation', 'platform'], drop_first=False)

sm_le_target = LabelEncoder()
sm_y = sm_le_target.fit_transform(sm_target_str)
sm_X = sm_features

print(f"\nX shape: {sm_X.shape}")
print(f"Признаки ({len(sm_X.columns)}): {list(sm_X.columns)}")
print(f"Классы таргета: {list(sm_le_target.classes_)}")
```

```python
# SM.2 Split + таргет-целевые counts для SMOTE
sm_X_train, sm_X_test, sm_y_train, sm_y_test = train_test_split(
    sm_X, sm_y, test_size=0.2, random_state=42, stratify=sm_y
)
print(f"train: {sm_X_train.shape[0]}, test: {sm_X_test.shape[0]}")
print(f"train class distribution: {dict(pd.Series(sm_y_train).value_counts())}")
print(f"test  class distribution: {dict(pd.Series(sm_y_test).value_counts())}")

# Для smmh классы уже относительно сбалансированы — оставим SMOTETomek с умеренным оверсемплингом
sm_majority = pd.Series(sm_y_train).value_counts().max()
sm_target_counts = {}
for cls, cnt in pd.Series(sm_y_train).value_counts().items():
    if cnt < sm_majority:
        sm_target_counts[int(cls)] = min(int(cnt * 2), int(sm_majority))
    else:
        sm_target_counts[int(cls)] = int(cnt)
print(f"\nЦелевое распределение для SMOTE: {sm_target_counts}")
```

```python
# SM.3 GridSearchCV с тем же Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

def sm_make_pipe(est):
    return ImbPipeline([
        ("scaler", StandardScaler()),
        ("smote", SMOTETomek(
            smote=SMOTE(sampling_strategy=sm_target_counts, random_state=42, k_neighbors=3),
            random_state=42)),
        ("clf", est),
    ])

sm_param_grids = {
    'Logistic Regression': (LogisticRegression(class_weight='balanced', max_iter=2000),
        {'clf__C': [0.01, 0.1, 1.0, 10.0], 'clf__solver': ['lbfgs', 'liblinear']}),
    'Decision Tree': (DecisionTreeClassifier(class_weight='balanced', random_state=42),
        {'clf__max_depth': [3, 5, 10, None], 'clf__min_samples_leaf': [1, 3, 5, 10]}),
    'Random Forest': (RandomForestClassifier(class_weight='balanced', random_state=42),
        {'clf__n_estimators': [100, 200], 'clf__max_depth': [5, 10, None],
         'clf__min_samples_leaf': [1, 3, 5]}),
    'SVM': (SVC(class_weight='balanced', probability=True, random_state=42),
        {'clf__C': [0.1, 1.0, 10.0], 'clf__kernel': ['rbf', 'linear'], 'clf__gamma': ['scale']}),
    'KNN': (KNeighborsClassifier(),
        {'clf__n_neighbors': [3, 5, 7, 11], 'clf__weights': ['uniform', 'distance'],
         'clf__metric': ['euclidean', 'manhattan']}),
}

sm_best_models = {}
sm_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
for name, (est, grid) in sm_param_grids.items():
    print(f"\n--- {name} ---")
    gs = GridSearchCV(sm_make_pipe(est), grid, cv=sm_cv, scoring='f1_macro', n_jobs=-1)
    gs.fit(sm_X_train, sm_y_train)
    sm_best_models[name] = {'pipeline': gs.best_estimator_,
                            'best_params': gs.best_params_,
                            'best_score': gs.best_score_}
    print(f"  Лучшие параметры: {gs.best_params_}")
    print(f"  CV F1-macro: {gs.best_score_:.4f}")
```

```python
# SM.4 Оценка на test
sm_results = []
sm_predictions = {}
for name, cfg in sm_best_models.items():
    pipe = cfg['pipeline']
    y_pred = pipe.predict(sm_X_test)
    try:
        y_proba = pipe.predict_proba(sm_X_test)
    except Exception:
        y_proba = None
    acc = float(np.mean(y_pred == sm_y_test))
    f1m = f1_score(sm_y_test, y_pred, average='macro')
    f1w = f1_score(sm_y_test, y_pred, average='weighted')
    prec = precision_score(sm_y_test, y_pred, average='macro', zero_division=0)
    rec = recall_score(sm_y_test, y_pred, average='macro', zero_division=0)
    roc = None
    if y_proba is not None:
        try: roc = roc_auc_score(sm_y_test, y_proba, multi_class='ovr', average='macro')
        except Exception: roc = None
    sm_results.append({
        'Model': name,
        'Accuracy': round(acc, 4),
        'Precision (macro)': round(prec, 4),
        'Recall (macro)': round(rec, 4),
        'F1 (macro)': round(f1m, 4),
        'F1 (weighted)': round(f1w, 4),
        'ROC-AUC (macro)': round(roc, 4) if roc else 'N/A',
    })
    sm_predictions[name] = y_pred
    print(f"\n=== {name} ===")
    print(classification_report(sm_y_test, y_pred, target_names=sm_le_target.classes_, zero_division=0))

sm_results_df = pd.DataFrame(sm_results).sort_values('F1 (macro)', ascending=False).reset_index(drop=True)
sm_results_df
```

## Нейронная сеть для smmh.csv

Тот же подход: `ImbPipeline(StandardScaler → SMOTETomek → KerasClassifier)` + `GridSearchCV` по сетке гиперпараметров. Сетка меньше, чем у синтетики (датасет маленький — смысла в глубокой сети нет).


```python
# SM.8 Neural network GridSearch на smmh
sm_n_features = sm_X_train.shape[1]
sm_n_classes = len(np.unique(sm_y_train))

def sm_create_model(n_hidden_layers=1, neurons=16, activation='relu', optimizer='adam'):
    model = Sequential()
    model.add(Input(shape=(sm_n_features,)))
    model.add(Dense(neurons, activation=activation, kernel_regularizer=l2(0.01)))
    model.add(Dropout(0.3))
    for _ in range(n_hidden_layers - 1):
        model.add(Dense(neurons, activation=activation, kernel_regularizer=l2(0.01)))
        model.add(Dropout(0.3))
    model.add(Dense(sm_n_classes, activation='softmax'))
    model.compile(loss='sparse_categorical_crossentropy', optimizer=optimizer, metrics=['accuracy'])
    return model

sm_keras_clf = KerasClassifier(model=sm_create_model, verbose=0)
sm_nn_pipe = ImbPipeline([
    ("scaler", StandardScaler()),
    ("smote", SMOTETomek(
        smote=SMOTE(sampling_strategy=sm_target_counts, random_state=42, k_neighbors=3),
        random_state=42)),
    ("clf", sm_keras_clf),
])

sm_param_grid_nn = {
    'clf__model__n_hidden_layers': [1, 2, 3],
    'clf__model__neurons': [16, 32, 64],
    'clf__model__activation': ['relu', 'tanh'],
    'clf__model__optimizer': ['adam', 'rmsprop'],
    'clf__batch_size': [16, 32],
    'clf__epochs': [50],
}

# n_jobs=-1 + verbose=2 — joblib показывает сводный прогресс
sm_grid_nn = GridSearchCV(
    estimator=sm_nn_pipe, param_grid=sm_param_grid_nn,
    cv=3, scoring='f1_macro',
    n_jobs=-1, verbose=2, pre_dispatch='2*n_jobs'
)
sm_grid_nn_result = sm_grid_nn.fit(sm_X_train, sm_y_train)

print(f"\nЛучший F1 (macro) на CV: {sm_grid_nn_result.best_score_:.4f}")
print(f"Лучшие параметры: {sm_grid_nn_result.best_params_}")
sm_best_nn_pipeline = sm_grid_nn_result.best_estimator_
```

```python
# SM.9 Оценка NN на test, добавление в общую таблицу результатов smmh
sm_y_pred_nn = sm_best_nn_pipeline.predict(sm_X_test)
try:
    sm_y_proba_nn = sm_best_nn_pipeline.predict_proba(sm_X_test)
except Exception:
    sm_y_proba_nn = None

sm_acc_nn = float(np.mean(sm_y_pred_nn == sm_y_test))
sm_f1m_nn = f1_score(sm_y_test, sm_y_pred_nn, average='macro')
sm_f1w_nn = f1_score(sm_y_test, sm_y_pred_nn, average='weighted')
sm_prec_nn = precision_score(sm_y_test, sm_y_pred_nn, average='macro', zero_division=0)
sm_rec_nn = recall_score(sm_y_test, sm_y_pred_nn, average='macro', zero_division=0)
sm_roc_nn = None
if sm_y_proba_nn is not None:
    try:
        sm_roc_nn = roc_auc_score(sm_y_test, sm_y_proba_nn, multi_class='ovr', average='macro')
    except Exception:
        sm_roc_nn = None

print(f"=== Neural Network on smmh ===")
print(classification_report(sm_y_test, sm_y_pred_nn,
      target_names=sm_le_target.classes_, zero_division=0))

# Добавим в общую таблицу sm_results и обновим лучшего
sm_results.append({
    'Model': 'Neural Network',
    'Accuracy': round(sm_acc_nn, 4),
    'Precision (macro)': round(sm_prec_nn, 4),
    'Recall (macro)': round(sm_rec_nn, 4),
    'F1 (macro)': round(sm_f1m_nn, 4),
    'F1 (weighted)': round(sm_f1w_nn, 4),
    'ROC-AUC (macro)': round(sm_roc_nn, 4) if sm_roc_nn else 'N/A',
})
sm_predictions['Neural Network'] = sm_y_pred_nn
sm_best_models['Neural Network'] = {
    'pipeline': sm_best_nn_pipeline,
    'best_params': sm_grid_nn_result.best_params_,
    'best_score': sm_grid_nn_result.best_score_,
}

sm_results_df = pd.DataFrame(sm_results).sort_values('F1 (macro)', ascending=False).reset_index(drop=True)
print("\n=== Итоговая таблица smmh (6 моделей) ===")
print(sm_results_df.to_string(index=False))
```

# Итоговые выводы

> **Характер работы.** Это **исследовательская/методологическая** работа на двух датасетах: (1) синтетическом `mental_health_social_media_dataset.csv` с детерминированной по конструкции целевой переменной и (2) реальном по природе `smmh.csv` без готовой клинической разметки — таргет на нём построен **эвристически** из пяти Likert-индикаторов дистресса. Цель работы — продемонстрировать корректность ML-пайплайна (EDA, предобработка, балансировка без утечки, GridSearchCV, сравнение моделей, интерпретация), а **не** получить клинически применимую модель прогноза психического состояния.

## 1. Прогон №1 — синтетический датасет

Train ≈ 3765 / test ≈ 942. GridSearchCV через `ImbPipeline` (5-fold, F1-macro). Категориальные признаки кодируются через `pd.get_dummies`. Балансировка SMOTETomek с `MAX_OVERSAMPLE_RATIO=3`.

Метрики 6 моделей (LR, DT, RF, SVM, KNN, NN) — см. таблицу в разделе «Сводная таблица метрик» и сравнительную таблицу ниже.

**Почему метрики на синтетике высокие** — задокументировано тремя независимыми проверками:
- **Feature importance (RF):** топ-5 признаков дают ~70–80% важности (`stress_level`, `anxiety_level`, `mood_level`, `sleep_hours`, `physical_activity_min`).
- **Learning curve:** F1-macro выходит на плато заметно раньше полного train-объёма — модели достаточно небольшой выборки, чтобы выучить формулу.
- **Ablation:** при удалении топ-3 признаков F1 падает существенно — топ-фичи фактически кодируют целевую переменную.

**Вывод:** в синтетическом датасете `mental_state` — детерминированная функция нескольких признаков. Это свойство **исходных данных**, а не качество пайплайна. Идеальные метрики на нём — артефакт конструкции датасета, а не доказательство модели.

## 2. Прогон №2 — реальные данные `smmh.csv` с эвристическим псевдо-таргетом

В smmh.csv нет готовой целевой переменной `mental_state`. Псевдо-таргет построен эвристически:
`score = q_restless + q_worries + q_concentrate + q_depressed + q_sleep_issues`,
`Healthy` если `score ≤ 10`, `At_Risk` если `score ≤ 17`, `Stressed` иначе.

Признаки — 13 колонок, **не пересекающиеся с формулой таргета** (демография, время в соцсетях, платформа, поведенческие Likert-вопросы про отвлекаемость / сравнение / валидацию / интерес). Тот же `ImbPipeline` (StandardScaler + SMOTETomek + классификатор), `pd.get_dummies` для категорий, GridSearchCV (5-fold, F1-macro).

Метрики 6 моделей — см. таблицу `sm_results_df`. Лучшие модели по CV F1-macro, test F1-macro и test Accuracy указаны отдельно во footnote-таблице (раздел «Сравнение двух прогонов»), чтобы не смешивать критерии в одной строке.

**Feature importance на smmh (RF):** важности распределены более равномерно; ни один признак не доминирует. В топе — поведенческие переменные (отвлекаемость, использование соцсетей без цели, сравнение с другими).

## 3. Сравнение двух прогонов

См. главную таблицу в ячейке 91. В одной строке приводится одна выбранная модель (по CV F1-macro) и **все её метрики** — без смешивания «лучшая модель по CV + лучшая F1 другой модели + лучшая accuracy третьей». Под главной таблицей — footnote с победителями по разным критериям для прозрачности.

**Что показывают цифры:**
- Метрики на синтетике близки к 1.0 — детерминированный таргет.
- Метрики на smmh.csv ~0.6–0.7 — пайплайн способен **восстанавливать собственный эвристический псевдо-таргет** на реальных по природе фичах. Это **не** клиническая валидация.
- Перенос модели, обученной на синтетике, на smmh даёт низкие метрики — закономерно, потому что распределения признаков и устройство таргета на двух датасетах принципиально разные.

## 4. Ограничения и характер работы

1. **Работа исследовательская/методологическая.** Не претендует на клиническую применимость прогноза `mental_state`.

4. **Для практического применения** (например, медицинский скрининг) необходим датасет с разметкой `mental_state`, выставленной специалистом по принятой клинической шкале (PHQ-9, GAD-7 и т. п.). Такого открытого датасета под эту конкретную задачу не нашлось — это объективное ограничение, не недочёт пайплайна.
